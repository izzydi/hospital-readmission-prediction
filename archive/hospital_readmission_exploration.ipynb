# Train-Test split and mode imputation

```r
# train-Test split 

# set seed for reproducibility

seed = 2023
set.seed(seed)


# initial train-test split with stratification

df_split <- 
  initial_split(
  df,
  prop = .75, # we keep 75% for the train set and 25% for the test set
  strata = Target)



# training and test sets

train_set <- training(df_split)
test_set <- testing(df_split)


# basic recipe with target as the response variable

basic_recipe <- recipe(Target ~ .,data = train_set)
 rcp <- basic_recipe %>%
 step_relevel(Target, ref_level = "yes") %>% # Relevel the target variable to "yes" for logistic regression
 step_impute_mode(all_nominal_predictors()) 
# Impute the mode for all nominal predictors with missing values.
# If we treat "Missing" as new category this imputation will not take place because we don't have any missing values.
# We don't need to remove it


# prepare and apply the recipe to the training set

df_train_imputed <-
 rcp %>%
 prep() %>%
 bake(new_data = NULL)

# apply the recipe to the test set

df_test_imputed <-
 rcp %>%
 prep() %>%
 bake(new_data = test_set)

# diagnose the data

df_train_imputed %>% 
 diagnose()
```

## Outliers 
<img src="hospital_4.jpg" alt="Drawing" style="height:310px;width:700px;"/>

Now is the time to address the issue of outliers in our data.Based on the box plots, it's evident that there are outliers present in the dataset. One of the most crucial steps of data preprocessing is detecting and handling outliers. But, what are outliers, in the first place? Simply to put the values that stand out of the crowd are outliers. For example: consider a list of ages of a sample of Netflix customers, (24, 34, 20, 16, 25, 30, 95, 21, 30, 15). Clearly 95 is an outlier.
Outliers can affect the model’s performance drastically. It is important to handle the outliers carefully. Boxplot is the best tool to visualize outliers in a dataset. Other ways are Histogram, Inter-Quartile Range and Z-score. Some ways to handle outliers are:

1. **Remove the outliers**: Before deleting the entries having outliers, make sure they aren’t adding any significant information and also that the dataset is not too small as deleting rows from datasets with a smaller number of records would mean losing vital information.
2. **Imputation with Mean/Median/Mode**: Replace the outliers with any of the measures of central tendency just as we do in missing value imputation.
3. **Quantile based flooring**: This method is like squeezing the far-off values so that they fall within the range.
4. **Transformation:** If the data is skewed, applying logarithmic, square or square root transformation might help.
5. **Regularization:** Regularization reduces variance and that makes it useful in handling outliers. The two most common types of regularization are Lasso and Ridge regularization, also known as L1 and L2 regularization respectively.
6. **Use tree based algorithms:** Instead of linear models, we can use tree-based methods like Random Forests and Gradient Boosting techniques, which are less impacted by outliers.
Let's create some additional plots to see how removing the outliers would affect the data.

```r
options(repr.plot.width = 8, repr.plot.height = 7, repr.plot.res = 100)
df_train_imputed %>%
 plot_outlier()
```

```r
df_train_imputed %>% 
 diagnose_outlier()
```

It looks like that if we remove outliers, the mean of each variable may alter slightly. Typically, I don't prefer removing outliers unless it is evident that it is a mistake. In most cases, consulting with an expert in the field, in this case a doctor, or obtaining more information about the data we are working with can assist in determining which values are outliers and which are not. Currently, I am considering imputing the outliers and utilizing a tree-based algorithm such as Random Forest to construct the model. I will run a GLM model for each variable to determine which imputation method to keep based on the AIC criterion.By comparing the AIC values of the models that are fit using each imputation method, we can select the approach that results in the best model fit. The method that results in the lowest AIC value can be considered the best approach for handling outliers in our data, at least in terms of model fit.However, it's important to note that selecting a method based solely on model fit may not be appropriate in all cases.That is why i wil also keep a dataframe with the outliers untouched, so we can assess how much the outliers are affecting our analysis and decide whether or not to remove or impute them based on their impact on our results. Below we see the AIC for each imputation method i am thinking to use.

```r
# predictor variables

vars <- c("time_in_hospital", "n_procedures", 
          "n_lab_procedures", "n_medications", 
          "n_outpatient", "n_inpatient", "n_emergency")

# imputation methods

impute_methods <- c("mean", "mode", "capping", "median")

# list to store models

models <- list()

# loop for all predictor variables

for (var in vars) {
  cat(paste0("--- ", var, " ---\n"))
    
  # fit a glm model without imputation
  dont_do_anything <- glm(Target ~ ., family = "binomial", 
                      data = df_train_imputed[, c(var, "Target")]) # we are selecting all rows, but only the columns named var and Target.
    
  cat("AIC (no imputation): ", round(AIC(dont_do_anything), 2), "\n")
    
  # loop over all imputation methods
  models[[var]] <- map(impute_methods, function(x) { 
 # models[[var]] will return the model that was fitted for the current predictor variable using the imputation method that was last looped over.
    df_train_imputed_method <- df_train_imputed %>%
      mutate({{ var }} := as.numeric(imputate_outlier(., {{ var }}, method = x))) 
      # The {{ var }} syntax is used to tell dplyr to 
      # take the value of var as a variable name, instead of interpreting it as a string.
      # The := operator is often used in conjunction with mutate() function to add or modify columns in a data frame.
    
    output <- glm(Target ~ ., family = "binomial", 
              data = df_train_imputed_method[, c(var, "Target")])
    
    cat(paste0("AIC (", x, "): ", round(AIC(output), 2), "\n"))
    
    return(output)
  })
}
```

I believe we can now choose the best imputation method based on the lowest AIC. 

```r
# data with outliers

df_with_outliers <- df_train_imputed 


# impute outliers

df_train_imputed <- df_train_imputed %>%
  mutate(
           time_in_hospital = imputate_outlier(df_train_imputed, time_in_hospital, method = "mode"),
           n_lab_procedures = imputate_outlier(df_train_imputed, n_lab_procedures, method = "mode"),
           n_procedures = imputate_outlier(df_train_imputed, n_procedures, method = "capping"),
           n_medications = imputate_outlier(df_train_imputed, n_medications, method = "mode"),
           n_outpatient = imputate_outlier(df_train_imputed, n_outpatient, method = "mean"),
           n_inpatient = imputate_outlier(df_train_imputed, n_outpatient, method = "capping")
           ) %>%
  mutate(across(c(time_in_hospital, n_lab_procedures, n_procedures, n_medications, n_outpatient,n_inpatient), as.numeric))
# the last mutation is because the imputate_outlier() function changes the class of the mutated vars to:"imputation" and "numeric"
# and this causes some problems with other R functions.
```

I am now thinking about conducting chi-square tests between our categorical variables and the target group. The chi-square test is a statistical hypothesis test that evaluates the relationship between nominal and ordinal variables in a bivariate table. In simpler terms, it checks if two variables are independent of one another. The null hypothesis assumes no relationship between the two variables, while the alternative hypothesis assumes that there is a relationship.

```r
options(warn = -1)  # supresswarnings

# function to perform chi-squared test and print the results

chisq_summary <- function(df, var) {
  counts <- table(df[[var]], df$Target)
  chisq.test(counts) %>%
    glance() %>%
    mutate(variable = var)
}

# names of all categorical variables in df_train_imputed

cat_vars <- df_train_imputed  %>% 
 select_if(is.factor)  %>% 
 select(-Target)%>%
 names()

# we apply the chisq_summary() function to each categorical variable and combine the results

chisq_results <- cat_vars %>% 
 map(~ chisq_summary(df_train_imputed, .x)) %>% 
 bind_rows()

# and we print the results 

chisq_results %>% 
  select(variable, p.value, statistic) %>%
  mutate(p.value = format(round(p.value, 3), nsmall = 3),
       statistic = format(round(statistic, 3), nsmall = 3))
```

The results of the chi-square test show that there is a significant relationship between all of the categorical variables and the target group. This suggests that there is a connection between the categorical variables and the target group.

# Anova tuning
To improve the performance of our models, we will tune them based on the ROC AUC metric. ROC AUC stands for Receiver Operating Characteristic Area Under the Curve, which is a widely used evaluation metric for binary classification problems. It is a measure of the model's ability to distinguish between the positive and negative classes across all possible thresholds of the predicted probabilities.

In ROC AUC, the true positive rate (TPR) is plotted against the false positive rate (FPR) at different threshold values, and the area under this curve is calculated. A perfect classifier will have an ROC AUC score of 1, while a random classifier will have a score of 0.5. A score of 0 indicates that the classifier is perfectly wrong, meaning that it predicts the opposite class from the true class.

Tuning the models based on ROC AUC means that we will try to find the optimal hyperparameters, such as the number of trees, maximum depth, learning rate, regularization strength, or feature subset size, that maximize the ROC AUC score on a validation set. This will help us to select the best model that maximizes the ability to distinguish between the positive and negative classes, which is often important in real-world applications such as medical diagnosis or fraud detection. 

We will use the "tune_race_anova()". Tune race anova is a function in the "tidymodels" R package that implements a tuning method called "ANOVA feature selection with repeated k-fold cross-validation.

## Models and workflows

```r
# xgboost model spec

xgb_spec <- 
   boost_tree(
     mtry = tune(), # number of variables to sample as candidates at each split
     tree_depth = tune(), # maximum depth of each tree
     learn_rate = tune(), # step size shrinkage used in update to prevent overfitting
     loss_reduction = tune(), # amount of loss reduction required to make a further partition on a leaf node of the tree
     min_n = tune(), # minimum number of samples per tree node required to continue splitting
     sample_size = tune(), # fraction of observations to sample for each tree
     trees = tune() # number of trees to fit
) %>% 
     set_engine("xgboost") %>% # set the engine to xgboost
     set_mode("classification") # set the mode to classification

#  xgboost workflow 

xgb_wf <- 
  workflow() %>%
  add_recipe(new_recipe) %>% # add data preprocessing recipe
  add_model(xgb_spec) # add model specification

#  random forest model spec

rf_spec <- 
   rand_forest(
     mtry = tune(), # number of variables to sample as candidates at each split
     min_n = tune(), # minimum number of samples per tree node required to continue splitting
     trees = tune() # number of trees to fit
) %>% 
     set_engine("ranger",importance = "permutation") %>% # set the engine to ranger
     set_mode("classification") 

# random forest workflow 

rf_wf <- 
  workflow() %>%
  add_recipe(new_recipe) %>% 
  add_model(rf_spec) 


# SVM model spec 

svm_spec <- 
  svm_rbf(
    cost = tune(), # cost of constraints violation
    rbf_sigma = tune() # RBF kernel bandwidth parameter
  ) %>%
  set_engine("kernlab") %>% 
  set_mode("classification") 

# SVM workflow

svm_wf <- 
  workflow() %>%
  add_recipe(new_recipe) %>% 
  add_model(svm_spec) 

# Logistic regression model 

log_reg_spec <- 
  logistic_reg(
    penalty = tune(), # regularization method (L1, L2, or elastic net)
    mixture = tune() # elastic net mixing parameter (only used if penalty = "elastic")
  ) %>%
  set_engine("glmnet") %>% 
  set_mode("classification") 

# Logistic regression workflow

log_reg_wf <- 
  workflow() %>%
  add_recipe(new_recipe) %>% 
  add_model(log_reg_spec)


# neural network spec 

nn_spec <- 
  mlp(
    hidden_units = tune(),  # The number of neurons in each hidden layer. 
    penalty = 0, # The weight decay penalty to be applied to the weights in each layer. 
    epochs = tune() # The number of epochs to train the neural network for. 
  ) %>% 
  set_engine("nnet") %>% 
  set_mode("classification")

# nn worklfow 

nn_wf <- 
  workflow() %>%
  add_recipe(new_recipe) %>% 
  add_model(nn_spec)


# cross-validation folds 

stack_folds <- vfold_cv(train_set, strata = "Target", v = 5)

# workflow set 

no_pre <- 
  workflow_set(
    preproc = list(anova_tuned = new_recipe),
    models = list(
      RF = rf_spec,
      XGB = xgb_spec,
      SVM = svm_spec,
      NN = nn_spec,
      LOG_REG = log_reg_spec
    )
  )

# Combine all workflows into a single list

all_workflows <- bind_rows(
         no_pre
)
```

## Tune race anova
The "tune_race_anova()" function uses a combination of random search and anova feature selection to optimize the hyperparameters of our models. The function eliminates parameter combinations that are not performing well and selects the best combination of hyperparameters for the model. This helps improve the accuracy and performance of the model. Once the best model and hyperparameters are selected, the model can be re-trained with these hyperparameters to improve its performance. 

```r
# control object for the race algorithm

race_ctrl <- control_race(
  save_pred = TRUE,     # Save the final model predictions
  save_workflow = TRUE, # Save the entire workflow
  verbose_elim = TRUE,  # Print information about feature elimination
  verbose = FALSE       # Don't print additional information during tuning
)

# use the `workflow_map` function to perform a tuning grid search across all workflows
# turn on parallel processing
registerDoParallel()

race_results <- all_workflows %>%
  workflow_map(
   "tune_race_anova",   # Use the `tune_race_anova` function for tuning
    seed = 006,          # Set the random seed to ensure reproducibility
    resamples = stack_folds,  # Use the `stack_folds` object for cross-validation
    grid = 40,           # Use a 40-point grid for tuning
    metrics = metric_set(roc_auc,f_meas),  # Use the ROC AUC metric for evaluation
    control = race_ctrl  # Use the control object defined above
  )
```

We can visualize how the possible parameter combinations we tried did during the “race.”

```r
# collect evaluation metrics for all models in `race_results`

race_metrics <- collect_metrics(race_results, summarize = FALSE)

# total number of models evaluated

number_of_all_models <- nrow(race_metrics)
cat("Total number of models evaluated:", number_of_all_models, "\n")

# plot of the ROC AUC metric for each model

autoplot(
  race_results,
  rank_metric = "roc_auc",
  metric = "roc_auc",
  select_best = TRUE
) + 
  geom_text(aes(y = mean - 1/2, label = wflow_id), angle = 90, hjust = 1, size = 7) + 
  geom_point(size = 5) + 
  lims(y = c(-0.5, 1)) +
  theme(legend.position = "none") +
  tema2
```

# Modeling with outliers
As previously mentioned, I will also train and tune a model on the original dataset that contains outliers to compare its performance with our previous models. This will also provide additional insights into the robustness of our model and help us better understand how it performs under different conditions.

## Preprocess

```r
train_set <- df_with_outliers
test_set <- df_test_imputed

new_recipe <- recipe(Target ~ ., data = train_set) %>%
  # create new features to the data
  step_mutate(total_procedures = n_lab_procedures + n_procedures,
              n_visits = n_inpatient + n_outpatient + n_emergency) %>%
  step_mutate(diabetes_related_procedures = ifelse(n_procedures > 0 &
                                                      (diag_1 %in% c("Diabetes") |
                                                       diag_2 %in% c("Diabetes") |
                                                       diag_3 %in% c("Diabetes")), 1, 0)) %>%
  step_mutate(primary_diagnosis = ifelse(diag_1 %in% c("Diabetes"), "Diabetes", "other") %>% as.factor) %>%
  
  # remove the vars we used for feature engineering
  step_rm(n_inpatient,n_outpatient,n_emergency,n_procedures,n_lab_procedures,diag_1,diag_2,diag_3) %>%
  
  # use dummy vars because we want to identify which groups of patients affect readmission rates
  step_dummy(all_nominal_predictors()) %>%
  
  # Normalizing the data helps to make sure that all features are on a comparable scale
  step_normalize(all_numeric_predictors()) %>%
  
  # This step is used to identify and remove variables that have zero variance, i.e. variables that have the same value for all observations
  step_zv(all_numeric_predictors()) %>%
  
  # use correlation filter just in case we still some correlated vars in our dataset
  step_corr(all_numeric_predictors(), method = c("pearson"))
```

## XGB model

```r
# xgboost model spec

xgb_spec <- 
   boost_tree(
     mtry = tune(), # number of variables to sample as candidates at each split
     tree_depth = tune(), # maximum depth of each tree
     learn_rate = tune(), # step size shrinkage used in update to prevent overfitting
     loss_reduction = tune(), # amount of loss reduction required to make a further partition on a leaf node of the tree
     min_n = tune(), # minimum number of samples per tree node required to continue splitting
     sample_size = tune(), # fraction of observations to sample for each tree
     trees = tune() # number of trees to fit
) %>% 
     set_engine("xgboost") %>% # set the engine to xgboost
     set_mode("classification") # set the mode to classification

#  xgboost workflow 

xgb_wf <- 
  workflow() %>%
  add_recipe(new_recipe) %>% # add data preprocessing recipe
  add_model(xgb_spec) # add model specification
```

```r
stack_folds <- vfold_cv(train_set, strata = "Target")

xgb_rs <- tune_race_anova(
  xgb_wf,
  resamples = stack_folds,
  grid = 50,
  metrics = metric_set(roc_auc,f_meas),
  control = control_race(verbose_elim = TRUE)
)
```
